```python
from rectangle import Rectangle
from circle import Circle
from shape2dplotter import Shape2dPlotter
import matplotlib.pyplot as plt
```

## test of rectangle

```python
rectangle1 = Rectangle(1, 1)
rectangle2 = Rectangle(6, 16)
rectangle3 = Rectangle(7, 4)
```

```python
rectangle1.translate(2, 3)
rectangle2.translate(-1, 4)
print(rectangle1)
print(rectangle2)
print(rectangle3)

print(rectangle1 == rectangle3)
print(rectangle2 == rectangle3)
```

## Test of circle

```python
circle1 = Circle(radius=4, x=2, y=3)
circle1 = Circle(5)
circle2 = Circle(4)
circle3 = Circle(7)
```

```python
circle1.translate(1, 2)
circle2.translate(-3, 4)
print(circle1)
print(circle2)
print(circle3)

print (circle1 == circle2)
print (circle2 == circle3)
```
